### Module 3: Causal 'What-If' Counterfactual Scenario Simulator

This notebook demonstrates the **Causal Counterfactual Scenario Simulator** using our pre-trained `TransformerForecaster` model. The simulator enables patients and clinicians to explore hypothetical non-pharmacological interventions (*'What-If' Scenarios*) prior to meal ingestion.

#### Intervention Scenarios Simulated:
* **1. Baseline Meal (No Change):** Unmodified original meal and physical activity profile.
* **2. Reduce Carbohydrates by 50%:** Dietary intervention scaling carbohydrate intake down by 50% (\(\text{Carbs}_{\text{new}} = \text{Carbs}_{\text{orig}} \times 0.50\)).
* **3. +30 min Post-Meal Walk:** Physical activity intervention simulating brisk post-meal walking (elevating METs activity in the time-series stream).
* **4. Combined Intervention:** Simultaneous 50% carbohydrate reduction AND 30-minute post-meal walk.

#### 5. Summary & Decision Support Insights

##### Key Observations:
* **Carbohydrate Reduction (-50%):** Significantly lowers peak postprandial height (\(G_{\max}\)) and flattens the glycemic excursion slope.
* **Post-Meal Walking (+30m):** Accelerates glucose disposal into skeletal muscle, reducing recovery time back to the clinical Target Range ($70 - 180\text{ mg/dL}$).
* **Combined Strategy:** Provides maximum Time-in-Range (TIR) and highest Clinical Utility Score, preventing postprandial hyperglycemia.
